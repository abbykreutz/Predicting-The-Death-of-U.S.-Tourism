Basic Bar chart

```python
# Count by region and country
region_country_counts = revised_df.groupby(['World Region', 'Country of Origin']).size().unstack(fill_value=0)

# Plot stacked bar chart (top 10 countries per region)
plt.figure(figsize=(14, 8))
region_country_counts.plot(kind='bar', stacked=True, colormap='tab20', figsize=(14, 8))
plt.title('Country Distribution by World Region', fontsize=14, fontweight='bold')
plt.xlabel('World Region', fontsize=12)
plt.ylabel('Number of Records', fontsize=12)
plt.xticks(rotation=45, ha='right')
plt.legend(title='Country', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.grid(axis='y', alpha=0.3)
plt.tight_layout()
plt.show()
```
